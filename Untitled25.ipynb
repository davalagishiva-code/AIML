### Finding Uncommon Elements in Pandas Series

First, we'll create two pandas Series, `seriesA` and `seriesB`.
Then, we'll convert them to sets to easily find the uncommon elements using set operations.

```python
import pandas as pd

# Create seriesA
seriesA = pd.Series([1, 2, 3, 4, 5])

# Create seriesB
seriesB = pd.Series([2, 4, 6, 8, 10])

print("Series A:")
display(seriesA)
print("\nSeries B:")
display(seriesB)
```

```python
# Convert Series to sets for efficient comparison
setA = set(seriesA)
setB = set(seriesB)

# Find items present in seriesA but not in seriesB
not_common_in_B = setA - setB

# Find items present in seriesB but not in seriesA
not_common_in_A = setB - setA

# Combine results to get all uncommon items
all_uncommon_items = not_common_in_A.union(not_common_in_B)

print("Items not common in both Series:")
display(list(all_uncommon_items))
```

```python
import pandas as pd
df = pd.read_csv("/content/iris.csv")
df
```

```python
count=df["species"].value_counts()
count
```

```python
import matplotlib.pyplot as plt
plt.bar(count.index, count.values)
plt.xlabel("Species")
plt.ylabel("Count")
plt.title("Count of Each Species")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a histogram for petal_length
plt.figure(figsize=(8, 6))
sns.histplot(df['petal_length'], bins=15, kde=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot for petal_length vs. petal_width, colored by species
plt.figure(figsize=(10, 7))
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=df, s=80)
plt.title('Relationship between Petal Length and Petal Width by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Species')
plt.show()
```
