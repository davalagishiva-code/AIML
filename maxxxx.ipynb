```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.ensemble import StackingRegressor
```

```python
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('https://aicsvv.netlify.app/tips.csv')
df
```

```python
target=df['tip']
```

```python
train=df[['total_bill']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(target,train,test_size=0.30)
```

```python
model_1=LinearRegression()
model_2=xgb.XGBRegressor()
model_3=RandomForestRegressor()
```

```python
all_models = [('lr', model_1),('xgb', model_2),('rf', model_3)]
```

```python
final_model =StackingRegressor(estimators=all_models,final_estimator=LinearRegression())
```

```python
final_model.fit(y_train, x_train)
```

```python
pred_final = final_model.predict(y_test)
```

```python
print(mean_squared_error(y_test,pred_final))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import log_loss, mean_squared_error
```

```python
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
df=pd.read_csv('https://aicsvv.netlify.app/tips.csv')
df
```

```python
target=df['tip']
```

```python
train=df[['total_bill']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(target,train,test_size=0.30)
```

```python
model_1=LinearRegression()
model_2=xgb.XGBRegressor()
model_3=RandomForestRegressor()
```

```python
final_model=VotingRegressor(estimators=[('lr',model_1),('xgb',model_2),('rf',model_3)])
```

```python
final_model.fit(y_train, x_train)
```

```python
pred_final=final_model.predict(y_test)
```

```python
print(log_loss,pred_final)
```

```python
print(mean_squared_error(y_test,pred_final))
```
